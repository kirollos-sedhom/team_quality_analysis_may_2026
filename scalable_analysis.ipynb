```python
import pandas as pd
import glob
import os

def load_and_consolidate_data(folder_path):
    # Find all Excel files in the target directory
    all_files = glob.glob(os.path.join(folder_path, "*.csv"))
    
    df_list = []
    for file in all_files:
        temp_df = pd.read_csv(file)
        df_list.append(temp_df)
        
    # Stack them vertically into one master table
    master_df = pd.concat(df_list, ignore_index=True)
    
    # Convert 'session date' to proper datetime objects to extract the Month
    master_df['session date'] = pd.to_datetime(master_df['Review_Cycle'])
    master_df['Month'] = master_df['session date'].dt.to_period('M') # e.g., '2026-04', '2026-05'
    
    return master_df

# Example execution:
# df = load_and_consolidate_data('./monthly_reports/')
```

```python
df = load_and_consolidate_data(".")
```
